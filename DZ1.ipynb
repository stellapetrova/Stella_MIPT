# Задание 1. Обработка деления на ноль
Напишите программу, которая принимает два числа от пользователя и выводит результат их деления. Используйте обработку исключений, чтобы корректно обработать ситуацию, когда пользователь вводит 0 в качестве второго числа.

```python
# Ввод данных
a=int(input())
b=int(input())
# Код, который может вызвать исключение
try:
    a/b
except ZeroDivisionError:
    # Обработка исключения
    print("Деление на ноль")
else:
    print(a/b)
```

# Задание 2. Обработка некорректного ввода
Расширьте предыдущую программу, чтобы она также обрабатывала ситуацию, когда пользователь вводит строку вместо числа. Используйте несколько блоков except для обработки разных типов исключений.

```python
# Код, который может вызвать исключение
try:
    # Ввод данных
    a=int(input())
    b=int(input())
    # Деление и вывод
    result=a/b
    print(result)
except ValueError:
    # Обработка исключения
    print("Введены не числа")
except ZeroDivisionError:
    # Обработка исключения
    print("Деление на ноль")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
```

# Задание 3. Создание собственных исключений
Напишите программу, которая вычисляет сумму списка целых чисел. Создайте свои собственные классы исключений для обработки ситуаций, когда в списке есть хотя бы одно чётное или отрицательное число. Используйте оператор raise для генерации исключений.

```python
class CountNumberError(Exception):
    # Исключение для чисел меньше нуля
    pass
class EvenNegativeError(Exception):
    # Исключение для чисел четных и отрицательных
    pass

# Код, который может вызвать исключение
try:
    # Ввод данных
    n=int(input("Количество элементов списка n = "))
    if n<0:
        raise CountNumberError("Введено не корректное количество элементов списка")
    spisok=[]
    for i in range(n):
        el=int(input())
        if el<0 or el%2==0:
            raise EvenNegativeError("Введено четное или отрицательное число")
        spisok.append(el)
    result=sum(spisok)
    print("Сумма списка равна", result)
except ValueError:
    # Обработка исключения
    print("Введены не целые числа")
except IndexError:
    # Обработка исключения
    print("Выход за пределы индексации")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
```

# Задание 4. Обработка ошибок индексации
Напишите программу, которая принимает от пользователя индекс элемента списка и выводит значение этого элемента. Используйте обработку исключений для корректной обработки ситуаций, когда пользователь вводит индекс, выходящий за пределы списка.

```python
# Пример списка
spisok=['123','abc','SA','5s46af','8-9']
print(spisok)
try:    
    k=int(input("Введите индекс (0-4) элемента списка для печати k = "))
    print(spisok[k])
except IndexError:
    # Обработка исключения
    print("Выход за пределы индексации")
except ValueError:
    # Обрабатываем ошибку нечислового ввода
    print("ОШИБКА: это не целое число!")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
```

# Задание 6. Обработка ошибок импорта модулей
Напишите программу, которая импортирует модуль math и использует функцию sqrt() для вычисления квадратного корня числа, введённого пользователем. Используйте обработку исключений для корректной обработки ситуаций, когда модуль math не может быть импортирован или функция sqrt() не может быть вызвана для отрицательного числа.

```python
class NegativeError(Exception):
    # Исключение для отрицательных чисел
    pass

try:
    import math
    n=float(input("Введите число для вычисления корня n = "))
    if n<0:
        raise NegativeError("Нельзя вычислить корень отрицательного числа")
    result=math.sqrt(n)
    print(result)
except NegativeError as e:
    # Добавляем обработку собственного исключения
    print(f"Отрицательное число: {e}")
except ImportError:
    print("Ошибка импорта модуля")
except AttributeError:
    print("Такой метод не существует")
except ModuleNotFoundError:
    print("Некорректное имя модуля")
except NameError:
    print("Использовано неекорректное имя в программе")
except ValueError:
    # Обрабатываем ошибку нечислового ввода
    print("ОШИБКА: это не число!")
except Exception as e:
    print(f"Непредвиденная ошибка: {e}")
```
